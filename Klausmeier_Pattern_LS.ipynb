```python
def adjacency_from_corr(C, thr):
    A = (C > thr).astype(int)
    np.fill_diagonal(A, 0)
    return A
```

```python
file = "functional_networks_continuous_dynamic.h5"

corr_matrices = []
a_list = []
biomass = []

with h5py.File(file, "r") as f:

    a_raw = f["a_values"][:]
    biomass_raw = f["biomass_final"][:]

    for i, a in enumerate(a_raw):

        C = f[f"a_{i:03d}"]["corr_matrix"][:]

        if not np.isfinite(C).all():
            continue

        corr_matrices.append(C)
        a_list.append(a)
        biomass.append(biomass_raw[i])

corr_matrices = np.array(corr_matrices)
a_list = np.array(a_list)
biomass = np.array(biomass)

# order the arrays by a_list
idx = np.argsort(a_list)

a_list = a_list[idx]
biomass = biomass[idx]
corr_matrices = corr_matrices[idx]

a1 = 0.1455  # desert → spots
a2 = 0.24    # spots → labs
a3 = 0.305   # labs → gaps

def add_phase_regions(fig, a_min, a_max):
    phases = [
        (a_min, 0.1455, "Desert", "rgba(255, 200, 150, 0.40)"),
        (0.1455, 0.24,   "Spots",  "rgba(150, 200, 255, 0.40)"),
        (0.24,   0.305,  "Labs",   "rgba(150, 255, 200, 0.40)"),
        (0.305,  a_max,  "Gaps",   "rgba(200, 150, 255, 0.40)")
    ]
    
    for x0, x1, name, color in phases:
        fig.add_vrect(
            x0=x0, x1=x1,
            fillcolor=color,
            layer="below",
            line_width=0,
            annotation_text=name,
            annotation_position="top left"
        )
    
    # Transition lines
    for x in [0.1455, 0.24, 0.305]:
        fig.add_vline(x=x, line_width=2, line_dash="dash", line_color="black")
    
    return fig

# NETWORK METRICS

mean_corr = []
std_corr = []
spectral_radius = []

for C in corr_matrices:
    upper = C[np.triu_indices_from(C, k=1)]

    mean_corr.append(np.mean(upper))
    std_corr.append(np.std(upper))

    eigvals = np.linalg.eigvalsh(C)
    spectral_radius.append(np.max(eigvals))

mean_corr = np.array(mean_corr)
std_corr = np.array(std_corr)
spectral_radius = np.array(spectral_radius)
biomass = np.array(biomass)
```

```python
def compute_metrics(corr_matrices, percentile):

    clustering = []
    spectral_graph = []

    for C in corr_matrices:

        upper = C[np.triu_indices_from(C, k=1)]
        thr = np.percentile(upper, percentile)

        A = (C > thr).astype(int)
        np.fill_diagonal(A, 0)

        G = nx.from_numpy_array(A)

        clustering.append(nx.average_clustering(G))

        eigvals = np.linalg.eigvalsh(A)
        spectral_graph.append(np.max(eigvals))

    return np.array(clustering), np.array(spectral_graph)
```

```python
thr_list = [70, 75, 80, 85, 90, 95]

results = {}

for thr in thr_list:
    c, s = compute_metrics(corr_matrices, thr)
    results[thr] = {
        "clustering": c,
        "spectral": s
    }
```

```python
# Trying other metrics to evaluate criticality: spectral gap, spectral entropy and participation ratio
spectral_gap = []
spectral_entropy = []
participation_ratio = []

for C in corr_matrices:
    eigvals = np.linalg.eigvalsh(C)
    p = eigvals / np.sum(eigvals)
    D = 1 / np.sum(p**2)
    S = -np.sum(p * np.log(p + 1e-12))

    spectral_entropy.append(S)
    spectral_gap.append(eigvals[-1] - eigvals[-2])
    participation_ratio.append(D)
```

```python
# NETWORK METRICS
mean_corr = []
std_corr = []
spectral_radius = []

for C in corr_matrices:

    upper = C[np.triu_indices_from(C, k=1)]

    mean_corr.append(np.mean(upper))
    std_corr.append(np.std(upper))

    eigvals = np.linalg.eigvalsh(C)
    spectral_radius.append(np.max(eigvals))


mean_corr = np.array(mean_corr)
std_corr = np.array(std_corr)
spectral_radius = np.array(spectral_radius)
biomass = np.array(biomass)

# ROUGHNESS / LOCAL FLUCTUATIONS
def calculate_roughness(x, y, window=5):

    # derivative
    dydx = np.gradient(y, x)

    # local variance of derivative
    derivative_variance = (
        pd.Series(dydx)
        .rolling(
            window=2*window+1,
            center=True
        )
        .var()
        .to_numpy()
    )

    # moving standard deviation
    moving_std = (
        pd.Series(y)
        .rolling(
            window=2*window+1,
            center=True
        )
        .std()
        .to_numpy()
    )

    return derivative_variance, moving_std

# Calculate roughness for every metric
mean_corr_deriv_var, mean_corr_window_std = calculate_roughness(
    a_list,
    mean_corr,
    window=3
)

std_corr_deriv_var, std_corr_window_std = calculate_roughness(
    a_list,
    std_corr,
    window=3
)

spectral_deriv_var, spectral_window_std = calculate_roughness(
    a_list,
    spectral_radius,
    window=3
)

biomass_deriv_var, biomass_window_std = calculate_roughness(
    a_list,
    biomass,
    window=3
)

# SELECT THRESHOLD 80%
clustering_80 = np.array(results[80]["clustering"])
spectral_80 = np.array(results[80]["spectral"])

# ROUGHNESS FUNCTION
def calculate_roughness(x, y, window=5):

    # derivative
    dydx = np.gradient(y, x)

    # variance of derivative in moving window
    derivative_variance = (
        pd.Series(dydx)
        .rolling(
            window=2*window+1,
            center=True
        )
        .var()
        .to_numpy()
    )

    # moving standard deviation of original curve
    moving_std = (
        pd.Series(y)
        .rolling(
            window=2*window+1,
            center=True
        )
        .std()
        .to_numpy()
    )

    return derivative_variance, moving_std

# APPLY METRICS
clustering_derivative_var, clustering_moving_std = calculate_roughness(
    a_list,
    clustering_80,
    window=5
)

spectral_derivative_var, spectral_moving_std = calculate_roughness(
    a_list,
    spectral_80,
    window=5
)

# CONVERT TO ARRAYS
spectral_entropy = np.array(spectral_entropy)
spectral_gap = np.array(spectral_gap)
participation_ratio = np.array(participation_ratio)

# ROUGHNESS FUNCTION
def calculate_roughness(x, y, window=5):

    # derivative
    dydx = np.gradient(y, x)

    # variance of derivative in moving window
    derivative_variance = (
        pd.Series(dydx)
        .rolling(
            window=2*window+1,
            center=True
        )
        .var()
        .to_numpy()
    )

    # moving standard deviation of original curve
    moving_std = (
        pd.Series(y)
        .rolling(
            window=2*window+1,
            center=True
        )
        .std()
        .to_numpy()
    )

    return derivative_variance, moving_std

# APPLY TO METRICS
entropy_derivative_var, entropy_moving_std = calculate_roughness(
    a_list,
    spectral_entropy,
    window=5
)


gap_derivative_var, gap_moving_std = calculate_roughness(
    a_list,
    spectral_gap,
    window=5
)


participation_derivative_var, participation_moving_std = calculate_roughness(
    a_list,
    participation_ratio,
    window=5
)
```

```python
# FUNCTION TO CALCULATE ROUGHNESS

def calculate_roughness(x, y, window=5):

    dydx = np.gradient(y, x)
    derivative_variance = (
        pd.Series(dydx)
        .rolling(
            window=2 * window + 1,
            center=True
        )
        .var()
        .to_numpy()
    )

    moving_std = (
        pd.Series(y)
        .rolling(
            window=2 * window + 1,
            center=True
        )
        .std()
        .to_numpy()
    )

    return derivative_variance, moving_std


# LOCALIZED STRUCTURE DATA
a_ls = np.array([a for a in sorted(ls_dynamic.keys()) if not is_excluded_a(a)])
B_before = np.array([ls_dynamic[a]["B_before"] for a in a_ls])
B_after = np.array([ls_dynamic[a]["B_after"] for a in a_ls])

# AFTER / BEFORE
after_before = B_after / (B_before + 1e-12)

# AFTER - BEFORE
after_before_difference = B_after - B_before

# ROUGHNESS OF AFTER / BEFORE
after_before_derivative_var, after_before_moving_std = (
    calculate_roughness(a_ls, after_before, window=5)
)

# ROUGHNESS OF AFTER - BEFORE
difference_derivative_var, difference_moving_std = (
    calculate_roughness(
        a_ls,
        after_before_difference,
        window=5
    )
)

# NORMALIZATION
def normalize_metric(y):

    y = np.asarray(y, dtype=float)
    normalized = np.full_like(y, np.nan)
    mask = np.isfinite(y)

    if np.sum(mask) == 0:
        return normalized

    y_min = np.min(y[mask])
    y_max = np.max(y[mask])

    if np.isclose(y_min, y_max):
        normalized[mask] = 0

    else:

        normalized[mask] = (y[mask] - y_min) / (y_max - y_min)

    return normalized


# METRICS
metrics_original = {
    "After / Before": after_before,
    "After - Before": after_before_difference,
    "Variance of derivative: After / Before": after_before_derivative_var,
    "Moving std: After / Before": after_before_moving_std,
    "Variance of derivative: After - Before": difference_derivative_var,
    "Moving std: After - Before": difference_moving_std,
    "Variance of derivative: Spectral Radius": spectral_derivative_var,
    "Moving std: Spectral Radius": spectral_moving_std,
    "Variance of derivative: Clustering": clustering_derivative_var,
    "Moving std: Clustering": clustering_moving_std,
    "Variance of derivative: Spectral Gap": gap_derivative_var,
    "Variance of derivative: Participation Ratio": participation_derivative_var
}


# NORMALIZED VERSION
metrics_normalized = {
    name: normalize_metric(values)
    for name, values in metrics_original.items()
}

fig = go.Figure()

# ADD ORIGINAL METRICS
for name, values in metrics_original.items():

    fig.add_trace(
        go.Scatter(
            x=a_ls,
            y=values,
            mode="lines+markers",
            name=name,
            visible=False,
            hovertemplate=(
                "<b>" + name + "</b><br>"
                "a = %{x:.5f}<br>"
                "Value = %{y:.6g}"
                "<extra></extra>"
            )
        )
    )


# ADD NORMALIZED METRICS
for name, values in metrics_normalized.items():

    fig.add_trace(
        go.Scatter(
            x=a_ls,
            y=values,
            mode="lines+markers",
            name=name,
            visible="legendonly",
            hovertemplate=(
                "<b>" + name + "</b><br>"
                "a = %{x:.5f}<br>"
                "Normalized value = %{y:.5f}"
                "<extra></extra>"
            )
        )
    )


# DEFAULT METRICS
default_metrics = [
    "After / Before",
    "After - Before",
    "Variance of derivative: Spectral Radius",
    "Moving std: Spectral Radius",
    "Variance of derivative: Clustering",
    "Moving std: Clustering",
    "Variance of derivative: Spectral Gap",
    "Variance of derivative: Participation Ratio"
]

metric_names = list(metrics_original.keys())
n_metrics = len(metric_names)


for i, name in enumerate(metric_names):

    if name in default_metrics:
        fig.data[
            n_metrics + i
        ].visible = True


# BUTTON VISIBILITY
visible_normalized = (
    [False] * n_metrics
    +
    [
        True
        if name in default_metrics
        else "legendonly"
        for name in metric_names
    ]
)

visible_original = (
    [
        True
        if name in default_metrics
        else "legendonly"
        for name in metric_names
    ]
    +
    [False] * n_metrics
)


# ADD PHASE REGIONS
fig = add_phase_regions(fig, np.min(a_ls), np.max(a_ls))

# FINAL LAYOUT
fig.update_layout(
    width=1400,
    height=750,
    title="Localized Recovery and Network Instability Metrics",
    xaxis={"title": "a",},
    yaxis={"title": "Normalized metric value"},
    template="simple_white",
    legend={
        "title": "Select metrics",
        "orientation": "v",
        "x": 1.02,
        "y": 1,
        "xanchor": "left",
        "yanchor": "top"
    },
    margin={"r": 450},
    updatemenus=[
        {
            "type": "buttons",
            "direction": "left",
            "x": 1,
            "y": 1.15,
            "xanchor": "right",
            "yanchor": "top",
            "buttons": [
                {
                    "label": "Normalized",
                    "method": "update",
                    "args": [
                        {
                            "visible": visible_normalized
                        },
                        {
                            "yaxis.title": "Normalized metric value"
                        }
                    ]
                },

                {
                    "label": "Original values",
                    "method": "update",
                    "args": [
                        {
                            "visible": visible_original
                        },
                        {
                            "yaxis.title": "Original metric value"
                        }
                    ]
                }

            ]

        }

    ]

)


fig.show()
```
